# Quotes and history
OpenTerm must be running (the NB command starts this notebook server next to it and sets `OPENTERM_URL`, which the cells pass to `OpenTerm()`).
Every DataFrame keeps where its data came from and when, in `df.attrs`.

```python
import os
from openterm_sdk import OpenTerm

ot = OpenTerm(os.environ.get("OPENTERM_URL", "http://127.0.0.1:8000"))
q = ot.quote(["AAPL", "MSFT", "SPY"])
q
```

```python
bars = ot.history("AAPL", "5Y", "1d")
print(bars.attrs["source"], bars.attrs["as_of_iso"], bars.attrs["stale"])
bars["close"].plot(title="AAPL close")
```
