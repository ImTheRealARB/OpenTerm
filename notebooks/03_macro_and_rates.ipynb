# Macro and rates
FRED series through QRY `DATES` mode (`FRED:<ID>` with the field `value`), and the Phase 30 boards as raw JSON.

```python
import os
import httpx
from openterm_sdk import OpenTerm

base = os.environ.get("OPENTERM_URL", "http://127.0.0.1:8000")
ot = OpenTerm(base)
curve = ot.query("QRY value FOR FRED:DGS2,FRED:DGS10 DATES -10Y..0 FREQ M")
(curve["FRED:DGS10 VALUE"] - curve["FRED:DGS2 VALUE"]).plot(title="10Y - 2Y, percentage points")
```

```python
mmr = httpx.get(f"{base}/api/macrod/mmr").json()
[(r["rate"], r["last"], r["date"]) for r in mmr["data"]["sections"][0]["rows"]]
```
