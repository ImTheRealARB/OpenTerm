# QRY from Python
`ot.query()` runs the same QRY the QRY panel runs. `ot.fields()` is FLD, the field catalog (source, unit, frequency, key per field).

```python
import os
from openterm_sdk import OpenTerm

ot = OpenTerm(os.environ.get("OPENTERM_URL", "http://127.0.0.1:8000"))
ot.fields("valuation")
```

```python
screen = ot.query("QRY last, pe, mcap, sector FOR MEMB:SPY WHERE mcap>2e11 SORT mcap LIMIT 25")
print(screen.attrs["notes"])
screen
```

```python
ts = ot.query("QRY close FOR SPY,TLT,FRED:DGS10 DATES -5Y..0 FREQ M")
ts.pct_change().corr()
```
